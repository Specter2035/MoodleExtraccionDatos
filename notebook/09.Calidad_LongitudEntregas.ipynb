```python
import json
import pandas as pd

# path inside Colab workspace
PATH = "grades_with_user_random.json"

# load json
with open(PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

# normalize into flat table
df = pd.json_normalize(data)

# basic cleaning
df["grade"] = pd.to_numeric(df["grade"], errors="coerce")
df["grade_randomized"] = df["grade_randomized"].astype(bool)

# optional: convert timestamps if present
for col in ["timecreated", "timemodified", "gradedtime"]:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], unit="s", errors="coerce")

# preview
print(df.shape)
df.head(200)
```
